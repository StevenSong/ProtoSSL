```python
import pandas as pd
from pathlib import Path

in_path = Path("/opt/gpudata/ecg/heedb")
out_path = Path("/opt/gpudata/ecg/heedb-spoof")
```

```python
mgh_meta = pd.read_csv(in_path / "I0001/metadata/metadata.csv", usecols=["BDSPPatientID", "SexDSC", "AgeAtAcquisition", "FileName"])
euh_meta = pd.read_csv(in_path / "I0006/metadata/metadata.csv", usecols=["BDSPPatientID", "Sex", "AgeAtAcquisition", "FileName"])
mgh_labels = pd.read_csv(in_path / "I0001/12SL_diagnoses/diagnoses_acquisition.csv")
euh_labels = pd.read_csv(in_path / "I0006/12SL_diagnoses/diagnoses_acquisition.csv")
```

```python
mgh_years = mgh_meta["FileName"].str.split("/").str[2]
euh_years = euh_meta["FileName"].str.split("/").str[1]
```

```python
mgh_years = mgh_years.astype(int)
euh_years = euh_years.astype(int)
```

```python
mgh_slim = mgh_meta[mgh_years.isin({2010, 2021, 2022})]
euh_slim = euh_meta[euh_years == 2010]
```

```python
mgh_slim_set = set(mgh_slim["FileName"])
euh_slim_set = set(euh_slim["FileName"])
```

```python
mgh_slim_labels = mgh_labels[mgh_labels["FileName"].isin(mgh_slim_set)]
euh_slim_labels = euh_labels[euh_labels["FileName"].isin(euh_slim_set)]
```

```python
mgh_slim.to_csv(out_path / "I0001/metadata/metadata.csv", index=False)
euh_slim.to_csv(out_path / "I0006/metadata/metadata.csv", index=False)
mgh_slim_labels.to_csv(out_path / "I0001/12SL_diagnoses/diagnoses_acquisition.csv", index=False)
euh_slim_labels.to_csv(out_path / "I0006/12SL_diagnoses/diagnoses_acquisition.csv", index=False)
```
